```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler, MaxAbsScaler
df = pd.read_csv(r"C:\Users\bhatt\Desktop\sample2.csv")
num = df.select_dtypes(include='number').columns
x = df[num]
result = pd.DataFrame({
 'original_mean': x.mean(),
 'minmax_mean': MinMaxScaler().fit_transform(x).mean(axis=0),
 'std_mean': StandardScaler().fit_transform(x).mean(axis=0),
 'robust_mean': RobustScaler().fit_transform(x).mean(axis=0),
 'maxabs_mean': MaxAbsScaler().fit_transform(x).mean(axis=0)
}, index=num)
print(result)
```
