```python
import pandas as pd, numpy as np
df = pd.read_csv(r"C:\Users\bhatt\Desktop\sample2.csv")
df_m=df.copy()
for col in df.select_dtypes(include='number').columns:
    Q1=df[col].quantile(0.25); Q3=df[col].quantile(0.75); IQR=Q3-Q1
    l,u=Q1-1.5*IQR, Q3+1.5*IQR
    df_m[col]=np.where((df[col]<l)|(df[col]>u), df[col].median(), df[col])
print(df_m.head())
```
