```python
import pandas as pd, numpy as np
df = pd.read_csv(r"C:\Users\bhatt\Desktop\sample2.csv")

# Binary Encoding manually (example: embarked column)
col = 'embarked'
uniq = df[col].astype(str).unique()
mapping = {v:i for i,v in enumerate(uniq)}
n_bits = len(bin(len(uniq)-1)) - 2 # kitne bits chahiye

for b in range(n_bits):
    df[f'{col}_bin_{b}'] = df[col].astype(str).map(mapping).apply(lambda x: (x >> b) & 1)

print(df.filter(like='embarked_bin').head())
```
