```python
import pandas as pd, matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
df = pd.read_csv(r"C:\Users\bhatt\Desktop\sample2.csv")
col = 'fare'
scaled = MinMaxScaler().fit_transform(df[[col]])
plt.hist(df[col].dropna(), bins=30, alpha=0.5, label='original')
plt.hist(scaled, bins=30, alpha=0.5, label='normalized')
plt.legend(); plt.show()
```
