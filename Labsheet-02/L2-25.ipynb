```python
import pandas as pd, matplotlib.pyplot as plt, seaborn as sns
from sklearn.preprocessing import StandardScaler
df = pd.read_csv(r"C:\Users\bhatt\Desktop\sample2.csv")
num = df.select_dtypes(include='number').columns
scaled = StandardScaler().fit_transform(df[num])
plt.figure(figsize=(12,5))
plt.subplot(1,2,1); sns.boxplot(data=df[num]); plt.title('Original'); plt.xticks(rotation=45)
plt.subplot(1,2,2); sns.boxplot(data=pd.DataFrame(scaled, columns=num)); plt.title('Scaled'); plt.xticks(rotation=45)
plt.tight_layout(); plt.show()
```
