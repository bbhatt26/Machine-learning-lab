```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
df = pd.read_csv(r"C:\Users\bhatt\Desktop\sample2.csv")
num = df.select_dtypes(include='number').columns
df_s = pd.DataFrame(MinMaxScaler().fit_transform(df[num]), columns=num)
print("Original:"); print(df[num].describe())
print("Normalized:"); print(df_s.describe())
```
