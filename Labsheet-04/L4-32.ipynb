```python
import os; os.environ['OMP_NUM_THREADS']='1'
import warnings; warnings.filterwarnings('ignore')
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
df = pd.read_csv(r"C:\Users\bhatt\Desktop\Mall_Customers.csv")
X = StandardScaler().fit_transform(df[['Age','Annual Income (k$)','Spending Score (1-100)']])
df['Cluster'] = KMeans(n_clusters=5, random_state=42, n_init=10).fit_predict(X)
print(df.groupby('Cluster')[['Age','Annual Income (k$)','Spending Score (1-100)']].mean())
print(df.groupby('Cluster').size())
```
