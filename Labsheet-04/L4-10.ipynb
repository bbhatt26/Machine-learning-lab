```python
import pandas as pd, matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
df = pd.read_csv(r"C:\Users\bhatt\Desktop\Mall_Customers.csv")
X = StandardScaler().fit_transform(df[['Age','Annual Income (k$)','Spending Score (1-100)']])
sse=[]
for k in range(1,11):
    sse.append(KMeans(n_clusters=k, random_state=42, n_init=10).fit(X).inertia_)
plt.plot(range(1,11), sse, marker='o')
plt.xlabel('K'); plt.ylabel('Inertia'); plt.title('Elbow Method')
plt.savefig('elbow.png')
print('saved elbow.png')
```
