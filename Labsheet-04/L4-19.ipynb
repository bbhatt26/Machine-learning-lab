```python
import warnings; warnings.filterwarnings('ignore')
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score
df = pd.read_csv(r"C:\Users\bhatt\Desktop\Mall_Customers.csv")
X = StandardScaler().fit_transform(df[['Age','Annual Income (k$)','Spending Score (1-100)']])
for link in ['ward','complete','average','single']:
    l = AgglomerativeClustering(n_clusters=5, linkage=link).fit_predict(X)
    print(link, silhouette_score(X, l))
```
