```python
import warnings; warnings.filterwarnings('ignore')
import pandas as pd, matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
df = pd.read_csv(r"C:\Users\bhatt\Desktop\Mall_Customers.csv")
X = StandardScaler().fit_transform(df[['Annual Income (k$)','Spending Score (1-100)']])
labels = AgglomerativeClustering(n_clusters=5, linkage='ward').fit_predict(X)
plt.scatter(X[:,0], X[:,1], c=labels, cmap='tab10')
plt.title('Hierarchical Clustering'); plt.savefig('hier.png'); print('saved')
```
