```python
import warnings; warnings.filterwarnings('ignore')
import pandas as pd, matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
df = pd.read_csv(r"C:\Users\bhatt\Desktop\Mall_Customers.csv")
X = StandardScaler().fit_transform(df[['Age','Annual Income (k$)','Spending Score (1-100)']])
Xp = PCA(n_components=2).fit_transform(X)
plt.scatter(Xp[:,0], Xp[:,1]); plt.xlabel('PC1'); plt.ylabel('PC2')
plt.title('PCA 2D'); plt.savefig('pca.png'); print('saved pca.png')
```
