```python
import warnings; warnings.filterwarnings('ignore')
import pandas as pd, matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
df = pd.read_csv(r"C:\Users\bhatt\Desktop\Mall_Customers.csv")
X = StandardScaler().fit_transform(df[['Age','Annual Income (k$)','Spending Score (1-100)']])
cum = PCA().fit(X).explained_variance_ratio_.cumsum()
print(cum)
plt.plot(range(1,4), cum, marker='o'); plt.savefig('cumvar.png'); print('saved')
```
