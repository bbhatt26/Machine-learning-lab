```python
import os; os.environ['OMP_NUM_THREADS']='1'
import warnings; warnings.filterwarnings('ignore')
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
df = pd.read_csv(r"C:\Users\bhatt\Desktop\Mall_Customers.csv")
X = StandardScaler().fit_transform(df[['Age','Annual Income (k$)','Spending Score (1-100)']])
Xp = PCA(n_components=2).fit_transform(X)
print(KMeans(n_clusters=5, random_state=42, n_init=10).fit_predict(Xp)[:10])
```
