```python
import pandas as pd, matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
df = pd.read_csv(r"C:\Users\bhatt\Desktop\Mall_Customers.csv")
X = StandardScaler().fit_transform(df[['Annual Income (k$)','Spending Score (1-100)']])
km = KMeans(n_clusters=5, random_state=42, n_init=10).fit(X)
plt.scatter(X[:,0], X[:,1], c=km.labels_, cmap='tab10')
plt.xlabel('Annual Income'); plt.ylabel('Spending Score'); plt.title('K-Means Clusters')
plt.savefig('clusters.png'); print('saved clusters.png')
```
